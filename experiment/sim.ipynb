```python
from Desim.Core import SimSession

from experiment.code_gen import HardwareConfig
from experiment.qwen_model_sim import SequenceConfig
from experiment.qwen_model_sim import NetworkConfig
from qwen_model_sim import ModelRuner,load_predefined_model_config
```

```python
model_name_list=[
    'Qwen2.5-7B-Instruct',
    'Qwen2.5-14B-Instruct',
    'Qwen2.5-32B-Instruct'
]
tp_config = {
    'Qwen2.5-7B-Instruct':1,
    'Qwen2.5-14B-Instruct':2,
    'Qwen2.5-32B-Instruct':4
}
```

```python
# 序列长度
# prefill 和 decode 分开计算

# prefill 的时间都是通用的
prefill_batch_size = 8
prefill_sequence_config_list = [
    SequenceConfig(batch_size=prefill_batch_size,prefill_length=128,decoding_length=0),
    # SequenceConfig(batch_size=4,prefill_length=512,decoding_length=0),
    SequenceConfig(batch_size=prefill_batch_size,prefill_length=2048,decoding_length=0),
]

# decode 实际上算的是总的，每个都是不同的
decode_sequence_config_list = [
    SequenceConfig(batch_size=1,prefill_length=128,decoding_length=128),
    # SequenceConfig(batch_size=1,prefill_length=128,decoding_length=512),
    SequenceConfig(batch_size=1,prefill_length=128,decoding_length=2048),

    # SequenceConfig(batch_size=1,prefill_length=512,decoding_length=128),
    # SequenceConfig(batch_size=1,prefill_length=512,decoding_length=512),
    # SequenceConfig(batch_size=1,prefill_length=512,decoding_length=2048),

    SequenceConfig(batch_size=1,prefill_length=2048,decoding_length=128),
    # SequenceConfig(batch_size=1,prefill_length=2048,decoding_length=512),
    SequenceConfig(batch_size=1,prefill_length=2048,decoding_length=2048),
]


```

```python

hardware_config = HardwareConfig(
    num_pim_unit= 16,
    pim_unit_id_list= [i for i in range(16)],
    pim_unit_sa_size= (8,128)
)

network_config = NetworkConfig(
    bandwidth=8,
    latency=100
)
```

```python
for model_name in model_name_list:
    model_config = load_predefined_model_config(model_name,r'D:\code\EdgeSim\experiment')
    for sequence_config in prefill_sequence_config_list:
        model_runner = ModelRuner(
            model_config = model_config,
            sequence_config = sequence_config,
            hardware_config = hardware_config,
            network_config=network_config,
            tp_size = tp_config[model_name],
            chunk_size=128,
        )

        model_latency = model_runner.get_model_latency()


        print(f'Model Name: {model_name} Batch Size: {sequence_config.
        batch_size} prefill_length: {sequence_config.prefill_length} decode_length: {sequence_config.decoding_length} Latency: {model_latency}\n')

```

```python
for model_name in model_name_list:
    model_config = load_predefined_model_config(model_name,r'D:\code\EdgeSim\experiment')
    for sequence_config in decode_sequence_config_list:
        model_runner = ModelRuner(
            model_config = model_config,
            sequence_config = sequence_config,
            hardware_config = hardware_config,
            network_config=network_config,
            tp_size = tp_config[model_name],
            chunk_size=128,
        )

        model_latency = model_runner.get_model_latency()


        print(f'Model Name: {model_name} Batch Size: {sequence_config.
        batch_size} prefill_length: {sequence_config.prefill_length} decode_length: {sequence_config.decoding_length} Latency: {model_latency}\n')
```
